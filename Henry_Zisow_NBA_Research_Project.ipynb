I believe that the taller an NBA player is, the lower their free throw percentage will be.

I want to know whether the height of an NBA player has an inverse relationship with their respective free throw percentage. I will collect data from basketball-reference.com and other websites that have NBA data. I will use the data to determine whether players' heights correlate to their free throw percentages.

```python
import re
import time
import requests
from bs4 import BeautifulSoup

# From Chapter 4
def all_numbers(text):
    return [float(n.strip(".").replace(",", "")) for n in re.findall(r"\d[\d,.]*", text)]

def first_number(text):
    found = all_numbers(text)
    return found[0] if found else None

print("Setup done.")
```

```python
def fetch_and_save(url, path):
  resp = requests.get(url, timeout=20)
  with open(path, "w", encoding="utf-8") as f:
    f.write(resp.text)
  return resp.status_code
```

```python
url = "https://www.basketball-reference.com/players/"
resp = requests.get(url, timeout=20)
resp.encoding = "utf-8"

print(resp.status_code)          # -> 200
print(len(resp.text))            # the whole page is one long string
print("Kareem" in resp.text)    # -> True
```

```python
BASE = "https://www.basketball-reference.com/players/"
player_first_letter_pages = []
alphabet = "abcdefghijklmnopqrstuvwxyz"
for letter in alphabet:
  player_first_letter_pages.append(f"{BASE}{letter}/")

print(player_first_letter_pages)
```

```python
files = []
for letter, page in zip(alphabet, player_first_letter_pages):
    path = f"{letter}.html"
    fetch_and_save(page, path)
    files.append(path)
    time.sleep(0.2)

print(len(files), "pages saved")
```

```python
player_ids = []
for letter in alphabet:
    with open(f"{letter}.html", encoding="utf-8") as f:
        text = f.read()
    pattern = r'data-append-csv="(.*?)" data-stat="player"'
    player_ids += re.findall(pattern, text)

print(len(player_ids))
print(player_ids[0])
```

```python
for player_id in player_ids[:2]:
    path = f"player_{player_id}.html"
    fetch_and_save(f"https://www.basketball-reference.com/players/{player_id[0]}/{player_id}.html", path)
    time.sleep(1)
```
